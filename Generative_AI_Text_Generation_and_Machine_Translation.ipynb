Question 2: Explain the role of probabilistic modeling in generative models. How do
these models differ from discriminative models?

  -> Probabilistic modeling plays a fundamental role in generative models by enabling them
to learn the underlying probability distribution of data. These models aim to capture
how data is generated by estimating the joint probability distribution P(X) or P(X, Y),
where X represents the input data and Y represents the labels. By learning this
distribution, generative models can generate new, realistic data samples that resemble
the training data.

In generative models, probabilistic modeling helps in:
- Understanding the structure and patterns present in the data.
- Generating new data points by sampling from the learned probability distribution.
- Handling uncertainty and variability in real-world data.
- Modeling complex and high-dimensional data distributions.

Examples of probabilistic generative models include:
- Naive Bayes
- Gaussian Mixture Models (GMM)
- Hidden Markov Models (HMM)
- Variational Autoencoders (VAE)
- Generative Adversarial Networks (GAN)

Difference Between Generative and Discriminative Models:

1. Objective:
   - Generative Models: Learn the joint probability distribution P(X, Y) or P(X) to
     generate data.
   - Discriminative Models: Learn the conditional probability P(Y | X) to directly
     perform prediction or classification.

2. Function:
   - Generative Models: Can generate new data samples and also perform classification.
   - Discriminative Models: Focus only on predicting output labels from input data.

3. Data Modeling:
   - Generative Models: Model how the data is produced.
   - Discriminative Models: Model decision boundaries between classes.

4. Examples:
   - Generative Models: Naive Bayes, GMM, HMM, VAE, GAN
   - Discriminative Models: Logistic Regression, Support Vector Machines (SVM),
     k-Nearest Neighbors (k-NN), Neural Networks


Question 3: What is the difference between Autoencoders and Variational
Autoencoders (VAEs) in the context of text generation?

  -> Autoencoders (AEs):
Autoencoders are neural network models designed to learn efficient compressed
representations (latent vectors) of input data by encoding the input into a lower-
dimensional space and then decoding it back to reconstruct the original input.
In text generation, autoencoders are mainly used for learning meaningful text
representations and reconstruction.

Key Characteristics:
- Learn a deterministic mapping from input text to latent space and back.
- Focus mainly on accurate reconstruction of the original text.
- Latent space is not continuous or well-structured for sampling.
- Limited ability to generate diverse and novel text.
- Mostly used for dimensionality reduction, denoising, and feature extraction.

Variational Autoencoders (VAEs):
VAEs are probabilistic generative models that learn a probability distribution
over the latent space instead of a fixed encoding. They model the latent variables
as distributions (usually Gaussian) and enable sampling from this distribution to
generate new data.

Key Characteristics:
- Learn a probabilistic mapping to a continuous and smooth latent space.
- Model latent variables as distributions rather than fixed values.
- Allow easy sampling for generating new and diverse text.
- Encourage smooth interpolation and better generalization.
- Suitable for creative and diverse text generation tasks.

Key Differences:

1. Latent Space:
   - Autoencoders: Deterministic and unstructured.
   - VAEs: Probabilistic, continuous, and structured.

2. Text Generation Ability:
   - Autoencoders: Poor at generating novel text; mainly reconstruct input.
   - VAEs: Good at generating new and diverse text samples.

3. Sampling:
   - Autoencoders: No meaningful sampling mechanism.
   - VAEs: Sample latent vectors from learned distributions for text generation.

4. Objective Function:
   - Autoencoders: Minimize reconstruction loss.
   - VAEs: Minimize reconstruction loss + KL divergence (regularization term).

5. Use Case in NLP:
   - Autoencoders: Feature learning, text compression, denoising.
   - VAEs: Generative text modeling, controlled text generation.


Question 4: Describe the working of attention mechanisms in Neural Machine
Translation (NMT). Why are they critical?
  
  -> Working of Attention Mechanisms in Neural Machine Translation (NMT):

In Neural Machine Translation, attention mechanisms allow the model to focus on
relevant parts of the source sentence while generating each word in the target
sentence. Instead of encoding the entire source sentence into a single fixed-length
vector, attention dynamically selects and weighs different words from the input
sequence during translation.

Working Steps:
1. The encoder processes the source sentence and produces a sequence of hidden states,
   each representing a word and its context.
2. During decoding, for each target word to be generated, the decoder computes
   attention scores between its current hidden state and all encoder hidden states.
3. These attention scores are normalized using a softmax function to obtain attention
   weights.
4. A context vector is computed as a weighted sum of encoder hidden states based on
   these attention weights.
5. The context vector is combined with the decoder state to generate the next target
   word.

This allows the model to dynamically align source words with target words during
translation, improving accuracy and fluency.

Why Attention Mechanisms are Critical:

1. Handles Long Sentences:
   - Prevents information loss caused by compressing long sequences into a single
     vector.

2. Improves Translation Quality:
   - Enables better word alignment between source and target languages.

3. Provides Interpretability:
   - Attention weights indicate which source words influenced each target word.

4. Enhances Context Awareness:
   - Allows the model to consider relevant words rather than the entire sentence.

5. Enables Transformer Models:
   - Attention is the core component of Transformer-based models, which outperform
     traditional RNN-based NMT systems.

```python
'''Question 6: Use the following small text dataset to train a simple Variational
Autoencoder (VAE) for text reconstruction:
["The sky is blue", "The sun is bright", "The grass is green",
"The night is dark", "The stars are shining"]
1. Preprocess the data (tokenize and pad the sequences).
2. Build a basic VAE model for text reconstruction.
3. Train the model and show how it reconstructs or generates similar sentences'''
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense, Lambda, Reshape
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras import backend as K

# --------------------
# Data
# --------------------
sentences = [
    "The sky is blue",
    "The sun is bright",
    "The grass is green",
    "The night is dark",
    "The stars are shining"
]

tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
sequences = tokenizer.texts_to_sequences(sentences)

max_len = max(len(s) for s in sequences)
x_train = pad_sequences(sequences, maxlen=max_len, padding="post")

vocab_size = len(tokenizer.word_index) + 1

# --------------------
# Sampling Layer
# --------------------
class Sampling(tf.keras.layers.Layer):
    def call(self, inputs):
        z_mean, z_log_var = inputs
        epsilon = tf.random.normal(shape=tf.shape(z_mean))
        return z_mean + tf.exp(0.5 * z_log_var) * epsilon

# --------------------
# Encoder
# --------------------
embedding_dim = 16
latent_dim = 8

encoder_inputs = Input(shape=(max_len,))
x = Embedding(vocab_size, embedding_dim)(encoder_inputs)
x = LSTM(32)(x)
z_mean = Dense(latent_dim)(x)
z_log_var = Dense(latent_dim)(x)
z = Sampling()([z_mean, z_log_var])

encoder = Model(encoder_inputs, [z_mean, z_log_var, z])

# --------------------
# Decoder
# --------------------
latent_inputs = Input(shape=(latent_dim,))
x = Dense(32, activation="relu")(latent_inputs)
x = Dense(max_len * vocab_size, activation="softmax")(x)
decoder_outputs = Reshape((max_len, vocab_size))(x)

decoder = Model(latent_inputs, decoder_outputs)

# --------------------
# VAE Model
# --------------------
class VAE(Model):
    def __init__(self, encoder, decoder, **kwargs):
        super().__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder

    def train_step(self, data):
        if isinstance(data, tuple):
            data = data[0]

        with tf.GradientTape() as tape:
            z_mean, z_log_var, z = self.encoder(data)
            reconstruction = self.decoder(z)

            recon_loss = tf.reduce_mean(
                tf.keras.losses.sparse_categorical_crossentropy(
                    data, reconstruction
                )
            )

            kl_loss = -0.5 * tf.reduce_mean(
                1 + z_log_var - tf.square(z_mean) - tf.exp(z_log_var)
            )

            total_loss = recon_loss + kl_loss

        grads = tape.gradient(total_loss, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))

        return {
            "loss": total_loss,
            "reconstruction_loss": recon_loss,
            "kl_loss": kl_loss,
        }

# Build and train
vae = VAE(encoder, decoder)
vae.compile(optimizer="adam")
vae.fit(x_train, epochs=300, verbose=0)

# --------------------
# Reconstruction
# --------------------
z_mean, _, z = encoder.predict(x_train)
reconstructed = decoder.predict(z)
predicted_sequences = np.argmax(reconstructed, axis=-1)

reverse_word_index = {v: k for k, v in tokenizer.word_index.items()}

def decode(seq):
    return " ".join([reverse_word_index.get(i, "") for i in seq if i != 0])

print("\nOriginal vs Reconstructed:\n")
for o, r in zip(x_train, predicted_sequences):
    print("Original     :", decode(o))
    print("Reconstructed:", decode(r))
    print("-" * 50)
```

```python
#Question 7: Use a pre-trained GPT model (like GPT-2 or GPT-3) to translate a short
#English paragraph into French and German. Provide the original and translated text.
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

def translate(text, model_name):
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

    inputs = tokenizer(text, return_tensors="pt", padding=True, truncation=True)
    outputs = model.generate(**inputs, max_length=128)

    return tokenizer.decode(outputs[0], skip_special_tokens=True)


english_text = (
    "Artificial Intelligence is transforming the world by enabling machines "
    "to learn, reason, and create. It is being used in healthcare, education, "
    "finance, and many other industries to improve efficiency and innovation."
)

french_translation = translate(english_text, "Helsinki-NLP/opus-mt-en-fr")
german_translation = translate(english_text, "Helsinki-NLP/opus-mt-en-de")

print("Original:\n", english_text)
print("\nFrench:\n", french_translation)
print("\nGerman:\n", german_translation)

```

```python
#Question 8: Implement a simple attention-based encoder-decoder model for English-to-Spanish translation using Tensorflow or PyTorch.
import tensorflow as tf
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# ----------------------------------
# Sample Dataset
# ----------------------------------
eng_sentences = [
    "hello",
    "how are you",
    "i love you",
    "good morning",
    "thank you"
]

spa_sentences = [
    "<start> hola <end>",
    "<start> como estas <end>",
    "<start> te amo <end>",
    "<start> buenos dias <end>",
    "<start> gracias <end>"
]

# ----------------------------------
# Tokenization
# ----------------------------------
eng_tokenizer = Tokenizer(filters='')
spa_tokenizer = Tokenizer(filters='')

eng_tokenizer.fit_on_texts(eng_sentences)
spa_tokenizer.fit_on_texts(spa_sentences)

eng_seq = eng_tokenizer.texts_to_sequences(eng_sentences)
spa_seq = spa_tokenizer.texts_to_sequences(spa_sentences)

max_eng_len = max(len(seq) for seq in eng_seq)
max_spa_len = max(len(seq) for seq in spa_seq)

eng_padded = pad_sequences(eng_seq, maxlen=max_eng_len, padding='post')
spa_padded = pad_sequences(spa_seq, maxlen=max_spa_len, padding='post')

eng_vocab_size = len(eng_tokenizer.word_index) + 1
spa_vocab_size = len(spa_tokenizer.word_index) + 1

# ----------------------------------
# Encoder
# ----------------------------------
class Encoder(tf.keras.Model):
    def __init__(self, vocab_size, embedding_dim, units):
        super().__init__()
        self.embedding = tf.keras.layers.Embedding(vocab_size, embedding_dim)
        self.lstm = tf.keras.layers.LSTM(units, return_sequences=True, return_state=True)

    def call(self, x):
        x = self.embedding(x)
        output, h, c = self.lstm(x)
        return output, h, c

# ----------------------------------
# Bahdanau Attention
# ----------------------------------
class Attention(tf.keras.layers.Layer):
    def __init__(self, units):
        super().__init__()
        self.W1 = tf.keras.layers.Dense(units)
        self.W2 = tf.keras.layers.Dense(units)
        self.V = tf.keras.layers.Dense(1)

    def call(self, enc_output, hidden):
        hidden = tf.expand_dims(hidden, 1)
        score = self.V(tf.nn.tanh(self.W1(enc_output) + self.W2(hidden)))
        attention_weights = tf.nn.softmax(score, axis=1)
        context = attention_weights * enc_output
        context = tf.reduce_sum(context, axis=1)
        return context, attention_weights

# ----------------------------------
# Decoder
# ----------------------------------
class Decoder(tf.keras.Model):
    def __init__(self, vocab_size, embedding_dim, units):
        super().__init__()
        self.embedding = tf.keras.layers.Embedding(vocab_size, embedding_dim)
        self.lstm = tf.keras.layers.LSTM(units, return_sequences=True, return_state=True)
        self.fc = tf.keras.layers.Dense(vocab_size)
        self.attention = Attention(units)

    def call(self, x, enc_output, hidden, cell):
        context, _ = self.attention(enc_output, hidden)
        x = self.embedding(x)
        x = tf.concat([tf.expand_dims(context, 1), x], axis=-1)
        output, h, c = self.lstm(x, initial_state=[hidden, cell])
        output = tf.reshape(output, (-1, output.shape[2]))
        output = self.fc(output)
        return output, h, c

# ----------------------------------
# Model Training
# ----------------------------------
embedding_dim = 64
units = 128
batch_size = 2
epochs = 300

encoder = Encoder(eng_vocab_size, embedding_dim, units)
decoder = Decoder(spa_vocab_size, embedding_dim, units)

optimizer = tf.keras.optimizers.Adam()
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)

def train_step(inp, targ):
    loss = 0
    with tf.GradientTape() as tape:
        enc_output, h, c = encoder(inp)
        dec_hidden = h
        dec_cell = c
        dec_input = tf.expand_dims([spa_tokenizer.word_index['<start>']] * inp.shape[0], 1)

        for t in range(1, targ.shape[1]):
            predictions, dec_hidden, dec_cell = decoder(
                dec_input, enc_output, dec_hidden, dec_cell
            )
            loss += loss_fn(targ[:, t], predictions)
            dec_input = tf.expand_dims(targ[:, t], 1)

    batch_loss = loss / targ.shape[1]
    variables = encoder.trainable_variables + decoder.trainable_variables
    gradients = tape.gradient(loss, variables)
    optimizer.apply_gradients(zip(gradients, variables))

    return batch_loss

# Training Loop
for epoch in range(epochs):
    loss = train_step(eng_padded, spa_padded)
    if epoch % 50 == 0:
        print(f"Epoch {epoch} | Loss {loss:.4f}")

# ----------------------------------
# Translation Function
# ----------------------------------
def translate(sentence):
    seq = eng_tokenizer.texts_to_sequences([sentence])
    seq = pad_sequences(seq, maxlen=max_eng_len, padding='post')

    enc_output, h, c = encoder(seq)
    dec_hidden, dec_cell = h, c
    dec_input = tf.expand_dims([spa_tokenizer.word_index['<start>']], 1)

    result = []
    for _ in range(max_spa_len):
        preds, dec_hidden, dec_cell = decoder(dec_input, enc_output, dec_hidden, dec_cell)
        pred_id = tf.argmax(preds[0]).numpy()
        word = spa_tokenizer.index_word.get(pred_id, '')
        if word == '<end>':
            break
        result.append(word)
        dec_input = tf.expand_dims([pred_id], 1)

    return " ".join(result)

# Test
print("\nTranslation Results:\n")
for s in eng_sentences:
    print(f"{s}  →  {translate(s)}")
```

```python
'''Question 9: Use the following short poetry dataset to simulate poem generation with a
pre-trained GPT model:
["Roses are red, violets are blue,",
"Sugar is sweet, and so are you.",
"The moon glows bright in silent skies,",
"A bird sings where the soft wind sighs."]
Using this dataset as a reference for poetic structure and language, generate a new 2-4
line poem using a pre-trained GPT model (such as GPT-2). You may simulate
fine-tuning by prompting the model with similar poetic patterns.
Include your code, the prompt used, and the generated poem in your answer.'''

from transformers import pipeline

# Load GPT-2 text generation pipeline
generator = pipeline("text-generation", model="gpt2")

# Given poetry dataset
dataset = [
    "Roses are red, violets are blue,",
    "Sugar is sweet, and so are you.",
    "The moon glows bright in silent skies,",
    "A bird sings where the soft wind sighs."
]

# Prompt (simulating fine-tuning)
prompt = (
    "Roses are red, violets are blue,\n"
    "Sugar is sweet, and so are you.\n"
    "The moon glows bright in silent skies,\n"
    "A bird sings where the soft wind sighs.\n\n"
    "Now write a new short poem in a similar style:\n"
)

# Generate poem
output = generator(
    prompt,
    max_length=80,
    num_return_sequences=1,
    temperature=0.8,
    top_k=50,
    top_p=0.95
)

generated_text = output[0]["generated_text"]

# Extract only the generated poem part
generated_poem = generated_text.replace(prompt, "").strip()

# Print results
print("Prompt Used:\n", prompt)
print("\nGenerated Poem:\n", generated_poem)
```
